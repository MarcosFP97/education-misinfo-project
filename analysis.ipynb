### CARGA DE DATOS

```python
import pandas as pd
import json 


### Etiquetador 1

e1 = pd.read_csv("./data/asesores_expertos/Etiquetador 1.csv")
e1['response_dict'] = e1['response'].apply(json.loads)

# Expandir a columnas
e1_expanded = pd.concat(
    [e1.drop(columns=['response']), e1['response_dict'].apply(pd.Series)],
    axis=1
)

# Opcional: eliminar la columna auxiliar
e1_expanded = e1_expanded.drop(columns=['response_dict'])
e1_expanded = e1_expanded[['image', 'credibility', 'quality', 'correctness', 'interest']]
print(len(e1_expanded))
print(e1_expanded.head())
```

```python
e2 = pd.read_csv("./data/asesores_expertos/Etiquetador 2.csv")
e2['response_dict'] = e2['response'].apply(json.loads)

# Expandir a columnas
e2_expanded = pd.concat(
    [e2.drop(columns=['response']), e2['response_dict'].apply(pd.Series)],
    axis=1
)

# Opcional: eliminar la columna auxiliar
e2_expanded = e2_expanded.drop(columns=['response_dict'])
e2_expanded = e2_expanded[['image', 'credibility', 'quality', 'correctness', 'interest']]
print(e2_expanded.head())
```

```python
e3 = pd.read_csv("./data/asesores_expertos/resultados_artai.csv")
e3['response_dict'] = e3['response'].apply(json.loads)

# Expandir a columnas
e3_expanded = pd.concat(
    [e3.drop(columns=['response']), e3['response_dict'].apply(pd.Series)],
    axis=1
)

# Opcional: eliminar la columna auxiliar
e3_expanded = e3_expanded.drop(columns=['response_dict'])
e3_expanded = e3_expanded[['image', 'credibility', 'quality', 'correctness', 'interest']]
print(e3_expanded.head())
```

```python
merged = e1_expanded.merge(e2_expanded, how="left", on=["image"]).merge(e3_expanded, how="left", on=["image"])
merged.head()
```

```python
merged
```

```python
def add_source(
    fn:str
):
    if "_x_" in fn:
        return "X"
    elif "claude" in fn or "chatgpt" in fn or "gemini" in fn:
        return "AI"
    return "news"

merged["source"] = merged["image"].map(add_source)
```

```python
merged
```

```python
result = (
    merged.groupby("source", group_keys=False)
      .apply(lambda g: g.nsmallest(7, "variance"))
)
```

### ANOTACIONES DE LOS ADOLESCENTES

```python
import pandas as pd

teens = pd.read_csv("resultados_combinados.csv")
teens.head()
```

```python
len(teens.groupby("sender")) ### 58 alumnos en total y cada uno anotó todas las imágenes, por lo que tenemos 58*21=1218 anotaciones. Un número de anotaciones muchísimo mayor que el de los expertos, que es 3*21=63. Esto nos permite tener una idea de la distribución de las anotaciones de los alumnos y compararlas con las de los expertos.
```

```python
import json 

teens['response_dict'] = teens['response'].apply(json.loads)

# Expandir a columnas
teens_expanded = pd.concat(
    [teens.drop(columns=['response']), teens['response_dict'].apply(pd.Series)],
    axis=1
)

# Opcional: eliminar la columna auxiliar
teens_expanded = teens_expanded.drop(columns=['response_dict'])
teens_expanded = teens_expanded[['sender', 'image', 'credibility', 'quality', 'correctness', 'interest']]
print(len(teens_expanded))
print(teens_expanded.head())
```

### ACUERDO POR DIMENSIÓN ENTRE LOS ADOLESCENTES

```python
#### credibilidad
import krippendorff
import numpy as np

individual_labels = {}
for _, group in teens_expanded.groupby("sender"):
    sender = group["sender"].iloc[0]
    if sender not in individual_labels:
        individual_labels[sender] = {}
    individual_labels[sender] = list(group["credibility"].values.astype(int).tolist())


data = np.array(list(individual_labels.values()))
# print(len(data[0]))
alpha = krippendorff.alpha(
    reliability_data=data,
    level_of_measurement='ordinal'
)
print(f'Krippendorff\'s alpha entre todos los adolescentes para la dimensión de credibilidad: {alpha}')
```
